## 2. CRISP-DM Framework Adaptation for ML Projects

```python

# Import required libraries for project planning and visualization
from datetime import datetime, timedelta
import matplotlib.pyplot as plt

print("🔄 CRISP-DM METHODOLOGY FOR ML PROJECTS")
print("=" * 42)

# Define CRISP-DM phases
scribe_crisp_dm_phases = {
    "1. Business Understanding": {
        "duration_weeks": 4,
        "activities": [
            "Define project objective: validate readability for target age",
            "Engage stakeholders (FCA, CIO, PM)",
            "Document business success criteria (MAE < 0.25)",
            "Conduct ethical and compliance scoping",
            "Agree development process, wireframes and user stories",
        ],
        "ml_specific_considerations": [
            "Transparency and explainability essential for public sector",
            "Regulatory alignment: GDPR, Equality Act",
            "Need for human-in-the-loop in production"
        ]
    },
    "2. Data Understanding": {
        "duration_weeks": 2,
        "activities": [
            "Collect CEFR and other text datasets",
            "Explore data quality, completeness and spread",
            "Visualise outliers and imbalance across age bands",
            "Evaluate Flesch-Kincaid alignment"
        ],
        "ml_specific_considerations": [
            "Verify dataset coverage for UK reading ages",
            "Detect potential data leakage",
            "Document text sampling assumptions"
        ]
    },
    "3. Data Preparation": {
        "duration_weeks": 2,
        "activities": [
            "Clean and re-band CEFR dataset",
            "Drop redundant or collinear metrics",
            "Split into training, test and gold standard sets",
            "Extract features using textstat and NLP tools"
        ],
        "ml_specific_considerations": [
            "Avoid reusing text across splits",
            "Ensure balance across age bands",
            "Support reproducibility with Git/versioning"
        ]
    },
    "4. Modeling": {
        "duration_weeks": 2,
        "activities": [
            "Train baseline Linear Regression",
            "Tune Ridge, Lasso, Tree-based and Ensemble models",
            "Use GridSearchCV for hyperparameter tuning",
            "Evaluate using MAE and interpretability"
        ],
        "ml_specific_considerations": [
            "Prioritise explainable models (e.g. Ridge)",
            "Visualise feature importances",
            "Document model assumptions and metrics"
        ]
    },
    "5. Evaluation": {
        "duration_weeks": 2,
        "activities": [
            "Test final models on unseen gold standard templates",
            "Compare predicted vs actual FK scores",
            "Validate business utility (age match)",
            "Prepare deployment recommendations"
        ],
        "ml_specific_considerations": [
            "Use MAE < 0.25 as success metric",
            "Test generalisability to new templates",
            "Ensure performance is stable and explainable"
        ]
    },
    "6. Deployment": {
        "duration_weeks": 4,
        "activities": [
            "Integrate into Azure DevOps pipeline",
            "Allow time to deploy and sign off in each environment (Dev, Test, UAT, Prod)",
            "Add alerts for MAE drift",
            "Document monitoring and retraining process",
            "Enable FCA user override and feedback loop"
        ],
        "ml_specific_considerations": [
            "Azure-native monitoring with Power BI",
            "Human-in-the-loop control for FCA edits",
            "Maintain audit trail and model versioning"
        ]
    }
}

# Calculate phase durations for Gantt-style timeline
start_date = datetime(2025, 4, 28)
current_date = start_date
phase_data = []

for phase_name, phase_info in scribe_crisp_dm_phases.items():
    duration = phase_info['duration_weeks']
    end_date = current_date + timedelta(weeks=duration)
    
    phase_data.append({
        'phase': phase_name,
        'start': current_date,
        'end': end_date,
        'duration': duration
    })
    
    current_date = end_date


# Prepare data for plotting
consistent_colors = ['yellow', 'pink', 'orange', 'red', 'green', 'cyan']


# Format labels as number \n title \n duration for better readability and space efficiency

fig, ax = plt.subplots(figsize=(14, 8))

for i, phase in enumerate(phase_data):
    bar_color = consistent_colors[i % len(consistent_colors)]
    bar_start = (phase['start'] - start_date).days / 7

    # Split phase name into number and title
    phase_parts = phase['phase'].split('. ', 1)
    label_number = phase_parts[0]
    label_title = phase_parts[1] if len(phase_parts) > 1 else phase['phase']
    label_text = f"{label_number}\n{label_title}\n({phase['duration']}w)"

    ax.barh(i, phase['duration'], left=bar_start, height=0.6,
            color=bar_color, edgecolor='black', alpha=0.9)

    # # Adjust text color based on bar width
    # if phase['duration'] >= 2:
    #     text_color = 'black'
    #     ha = 'center'
    # else:
    #     text_color = 'black'
    #     ha = 'left'

    # ax.text(bar_start + phase['duration'] / 2, i,
    #         label_text,
    #         ha=ha, va='center', fontsize=9, fontweight='bold', color=text_color,
    #         wrap=True)

ax.set_yticks(range(len(phase_data)))
ax.set_yticklabels([p['phase'] for p in phase_data], fontsize=14)
ax.set_xlabel('Weeks from Project Start', fontsize=14)
ax.set_title('Scribe ML Project Timeline (CRISP-DM Framework)', fontsize=18, fontweight='bold', pad=20)

ax.grid(True, which='major', axis='x', linestyle='--', alpha=0.3)
ax.set_facecolor('#f8f8f8')
fig.patch.set_facecolor('white')

plt.tight_layout()
plt.show()



# Display phase details
for phase_name, phase_info in scribe_crisp_dm_phases.items():
    print(f"\n📅 {phase_name.upper()}")
    print(f"Duration: {phase_info['duration_weeks']} weeks")
    print("Key Activities:")
    for activity in phase_info['activities']:
        print(f"  • {activity}")
    print("ML-Specific Considerations:")
    for consideration in phase_info['ml_specific_considerations']:
        print(f"  ⚡ {consideration}")
```

## 3. Risk Management for ML Projects

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from collections import defaultdict
from textwrap import fill

print("⚠️  RISK MANAGEMENT FRAMEWORK")
print("=" * 35)

# Adapted ML risks for the Scribe reading age validation project
scribe_ml_risks = {
    "Data Quality Risks": {
        "Unsuitable Reading Age Data": {
            "probability": "Medium",
            "impact": "High",
            "description": "Training data does not reflect UK-based child reading levels",
            "mitigation": [
                "Re-band CEFR dataset to UK reading standards",
                "Use gold standard template letters for validation",
                "Visualise and balance dataset by age group"
            ]
        },
        "Data Drift": {
            "probability": "High",
            "impact": "Medium",
            "description": "Text complexity in Scribe may evolve post-deployment",
            "mitigation": [
                "Implement MAE-based drift alerts",
                "Plan periodic retraining using new templates",
                "Log all predictions for auditability"
            ]
        },
        "Bias in Readability Scores": {
            "probability": "Medium",
            "impact": "High",
            "description": "Standard scores may not reflect children with SEND or EAL",
            "mitigation": [
                "Include FCA feedback loop (human-in-the-loop)",
                "Cross-validate using multiple readability metrics",
                "Flag texts for manual override"
            ]
        }
    },
    "Technical Risks": {
        "Model Generalisation Failure": {
            "probability": "Medium",
            "impact": "High",
            "description": "Model performs poorly on unseen or edge-case letters",
            "mitigation": [
                "Test on gold standard letters",
                "Use multiple models and ensemble methods",
                "Track performance over time with Power BI"
            ]
        },
        "Integration Complexity": {
            "probability": "High",
            "impact": "Medium",
            "description": "Difficulty integrating ML model into Azure-based pipeline",
            "mitigation": [
                "Follow existing Azure DevOps process",
                "Deploy via feature branches with approval gates",
                "Engage DevOps team early"
            ]
        },
        "Overfitting to Training Data": {
            "probability": "Medium",
            "impact": "Medium",
            "description": "Model performs well on training but poorly on validation",
            "mitigation": [
                "Use MAE as a robust metric",
                "Apply regularisation (Ridge, Lasso)",
                "Cross-validate with unseen gold standard data"
            ]
        }
    },
    "Business Risks": {
        "Low User Trust": {
            "probability": "Medium",
            "impact": "High",
            "description": "FCA staff may reject ML output if seen as unreliable or opaque",
            "mitigation": [
                "Provide interpretable model explanations",
                "Keep FCA-in-the-loop at all times",
                "Pilot and gather feedback before full rollout"
            ]
        },
        "Non-Compliance with Accessibility Standards": {
            "probability": "Low",
            "impact": "High",
            "description": "Generated letters fail to meet accessibility guidelines",
            "mitigation": [
                "Use Flesch-Kincaid and CEFR alignment",
                "Manually review templates in early stage",
                "Ensure WCAG 2.2 compliance where possible"
            ]
        },
        "Limited Business Value Realisation": {
            "probability": "Medium",
            "impact": "High",
            "description": "Tool fails to reduce FCA workload or improve clarity",
            "mitigation": [
                "Track time savings and workload impact",
                "Conduct comparative testing (manual vs ML)",
                "Phase rollout based on usage patterns"
            ]
        }
    }
}

# Plot risk matrix

colors = {
    'Data Quality Risks': 'red',
    'Technical Risks': 'orange',
    'Business Risks': 'blue'
}    

risk_matrix_data = []
for category, risks in scribe_ml_risks.items():
    for risk_name, risk_info in risks.items():
        prob_score = {"Low": 1, "Medium": 2, "High": 3}[risk_info["probability"]]
        impact_score = {"Low": 1, "Medium": 2, "High": 3}[risk_info["impact"]]
        
        risk_matrix_data.append({
            "Risk": risk_name,
            "Category": category,
            "Probability": prob_score,
            "Impact": impact_score,
            "Risk_Score": prob_score * impact_score,
            "Description": risk_info["description"]
        })

risk_df = pd.DataFrame(risk_matrix_data)

risk_positions = defaultdict(list)
for _, row in risk_df.iterrows():
    pos = (row['Probability'], row['Impact'])
    risk_positions[pos].append(row)

fig, ax = plt.subplots(figsize=(10, 8))
ax.set_xlim(0.5, 3.5)
ax.set_ylim(0.5, 3.5)

# Draw grid boxes
for x in [1, 2, 3]:
    for y in [1, 2, 3]:
        ax.add_patch(plt.Rectangle((x - 0.5, y - 0.5), 1, 1,
                                   edgecolor='black', facecolor='white', alpha=0.05))


for (x, y), risks in risk_positions.items():
    offset_step = 0.18
    start_y = y - (offset_step * (len(risks)-1) / 2)
    for i, row in enumerate(risks):
        wrapped_label = fill(row['Risk'], width=20)
        ax.text(x, start_y + i * offset_step, wrapped_label,
                ha='center', va='center', fontsize=8,
                bbox=dict(facecolor=colors[row['Category']], alpha=0.5, boxstyle="round,pad=0.3"))

# Axes setup
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(['Low', 'Medium', 'High'])
ax.set_yticks([1, 2, 3])
ax.set_yticklabels(['Low', 'Medium', 'High'])
ax.set_xlabel('Probability', fontsize=12)
ax.set_ylabel('Impact', fontsize=12)
ax.set_title('Scribe ML Risk Matrix', fontsize=14, fontweight='bold')
ax.grid(False)

# Legend
legend_elements = [
    Patch(facecolor='red', edgecolor='black', alpha=0.5, label='Data Quality Risks'),
    Patch(facecolor='orange', edgecolor='black', alpha=0.5, label='Technical Risks'),
    Patch(facecolor='blue', edgecolor='black', alpha=0.5, label='Business Risks')
]
ax.legend(handles=legend_elements, title='Risk Categories', loc='upper left', bbox_to_anchor=(1.02, 1))

plt.tight_layout()
plt.show()


# Display top risks
print("🚨 HIGH-PRIORITY RISKS:")
print("-" * 25)
top_risks = risk_df.nlargest(5, 'Risk_Score')
for _, risk in top_risks.iterrows():
    print(f"\n{risk['Risk']} (Score: {risk['Risk_Score']})")
    print(f"  Category: {risk['Category']}")
    print(f"  Description: {risk['Description']}")
    
    # Find mitigation strategies
    for category, risks in scribe_ml_risks.items():
        if risk['Risk'] in risks:
            print(f"  Mitigation Strategies:")
            for strategy in scribe_ml_risks[category][risk['Risk']]['mitigation']:
                print(f"    • {strategy}")
```

## 4. Resource Planning and Team Structure

```python
import numpy as np

print("👥 RESOURCE PLANNING AND TEAM STRUCTURE")
print("=" * 42)

# Full team definition with all 8 roles
full_team_structure = {
    "Core Team": {
        "Tech Lead": {"allocation": "100%", "duration": "16 weeks"},
        "AI Engineer/Frontend Developer": {"allocation": "100%", "duration": "16 weeks"},
        "Backend Developer": {"allocation": "100%", "duration": "16 weeks"},
        "QA Tester": {"allocation": "50%", "duration": "10 weeks"}
    },
    "Supporting Team": {
        "Project Manager": {"allocation": "100%", "duration": "16 weeks"},
        "Azure DevOps Engineer": {"allocation": "50%", "duration": "10 weeks"},
        "Business Analyst": {"allocation": "50%", "duration": "16 weeks"}
    }
}

hourly_rates = {
    "Project Manager": 625 / 7.5,
    "Tech Lead": 930 / 7.5,
    "AI Engineer/Frontend Developer": 650 / 7.5,
    "Backend Developer": 650 / 7.5,
    "QA Tester": 500 / 7.5,
    "Azure DevOps Engineer": 650 / 7.5,
    "Business Analyst": 650 / 7.5
}

roles, allocations, durations, costs = [], [], [], []

for category, members in full_team_structure.items():
    for role, details in members.items():
        roles.append(role)
        allocation = float(details["allocation"].strip('%')) / 100
        duration = int(details["duration"].split()[0])
        allocations.append(allocation)
        durations.append(duration)

        weekly_hours = 37.5 * allocation
        total_hours = weekly_hours * duration
        total_cost = total_hours * hourly_rates[role]
        costs.append(total_cost)

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

y_pos = np.arange(len(roles))
bars = ax1.barh(y_pos, durations, alpha=0.8)

core_count = len(full_team_structure["Core Team"])
for i, bar in enumerate(bars):
    bar.set_color('mediumseagreen' if i < core_count else 'orchid')

ax1.set_yticks(y_pos)
ax1.set_yticklabels(roles)
ax1.set_xlabel('Duration (weeks)')
ax1.set_title('Team Member Duration (Full Team)', fontweight='bold')
ax1.grid(True, alpha=0.3)

for i, (duration, allocation) in enumerate(zip(durations, allocations)):
    ax1.text(duration + 0.5, i, f'{allocation:.0%}', va='center', fontweight='bold')

ax2.pie(costs, labels=roles, autopct='%1.1f%%', startangle=90)
ax2.set_title('Cost Distribution by Role (Full Team)', fontweight='bold')

plt.tight_layout()
plt.show()

# Print cost summary
total_cost = sum(costs)
print(f"💰 FULL TEAM COST SUMMARY:")
print(f"Total Project Cost: £{total_cost:,.0f}")
print(f"Average Weekly Cost: £{total_cost/20:,.0f}")

print(f"\n📊 COST BREAKDOWN BY ROLE:")
for role, cost in zip(roles, costs):
    print(f"  {role}: £{cost:,.0f} ({(cost / total_cost) * 100:.1f}%)")
```

## 7. Change Management and Deployment Strategy

```python
import matplotlib.pyplot as plt

# Define change management framework (ADKAR model)
change_management = {
    "Awareness": {
        "timeline": "Weeks 1–4",
        "activities": [
            "Initial executive briefing to CIO and PM",
            "Kick-off with Cafcass and Version 1 teams",
            "Explain role of AI and ML in letter validation",
            "Ensure stakeholders understand purpose and ethics"
        ],
        "deliverables": [
            "Vision and scope document",
            "Slide deck on AI explainability and compliance",
            "Introductory stakeholder comms"
        ]
    },
    "Desire": {
        "timeline": "Weeks 3–6",
        "activities": [
            "Showcase mock-ups and expected impact on FCA workload",
            "Gather early feedback from FCA staff",
            "Address scepticism around readability scoring and OpenAI use",
            "Communicate how the tool aligns with Ofsted and accessibility goals"
        ],
        "deliverables": [
            "Feedback report",
            "Adoption barriers document",
            "Change champions identified in FCA teams"
        ]
    },
    "Knowledge": {
        "timeline": "Weeks 5–10",
        "activities": [
            "Train-the-trainer sessions for project leads and FCA advisors",
            "Write clear, simple documentation on model limitations and overrides",
            "Create a knowledge base (markdown/wiki/FAQs)"
        ],
        "deliverables": [
            "Training materials and recorded walkthroughs",
            "User handbook for letter validation dashboard",
            "GitHub wiki content"
        ]
    },
    "Ability": {
        "timeline": "Weeks 10–14",
        "activities": [
            "Run UAT and gather usability feedback",
            "Provide test data and simulate real case usage",
            "Introduce override workflow and model audit view"
        ],
        "deliverables": [
            "UAT sign-off document",
            "Performance evaluation results",
            "Support plan handover to Version 1 BAU"
        ]
    },
    "Reinforcement": {
        "timeline": "Weeks 14–20",
        "activities": [
            "Model performance dashboard live in Power BI",
            "Quarterly performance review meetings",
            "User feedback survey and roadmap check-in"
        ],
        "deliverables": [
            "MAE monitoring alerting",
            "Lessons learned report",
            "Proposed backlog items for next phase"
        ]
    }
}

# Define deployment strategy
deployment_strategy = {
    "Phase 1: Development": {
        "duration": "2 weeks",
        "scope": "Model developed and tested locally by developer team",
        "participants": "ML Engineer, Backend Developer, Tech Lead",
        "success_criteria": [
            "Codebase version-controlled and tested",
            "Preliminary MAE < 0.25 achieved on validation set",
            "Documentation of model assumptions and architecture"
        ],
        "rollback_plan": "Code remains in dev branch with no further promotion"
    },
    "Phase 2: Test": {
        "duration": "2 weeks",
        "scope": "Deployed to isolated test environment with synthetic/mock data",
        "participants": "QA Testers, ML Engineer, Tech Lead",
        "success_criteria": [
            "Model deploys cleanly in CI/CD pipeline",
            "Test results meet acceptance criteria (MAE and latency)",
            "Logging, monitoring, and rollback mechanisms validated"
        ],
        "rollback_plan": "Restore to last stable build in CI/CD"
    },
    "Phase 3: UAT": {
        "duration": "3 weeks",
        "scope": "Model tested by end users (FCA staff) with real historical letters",
        "participants": "Product Owner, Product Manager, FCA Users",
        "success_criteria": [
            "FCA feedback ≥ 4.0/5.0 satisfaction",
            "No major interpretability issues raised",
            "MAE < 0.25 confirmed on gold standard test set"
        ],
        "rollback_plan": "Feedback used to retrain or revise before Prod release"
    },
    "Phase 4: Production": {
        "duration": "4 weeks",
        "scope": "Full-scale deployment with real-time letter scoring",
        "participants": "Customer Services, DevOps, Product Owner",
        "success_criteria": [
            "99.5% uptime for scoring service",
            "Retraining pipeline in place with MAE monitoring",
            "Audit log and override system enabled"
        ],
        "rollback_plan": "Controlled revert to UAT configuration with rollback scripts"
    }
}

# Process change management timeline
def extract_weeks(timeline):
    timeline = timeline.replace("–", "-")
    if "Weeks" in timeline:
        timeline = timeline.split("Weeks ")[1]
    if "-" in timeline:
        start, end = timeline.split('-')
        return int(start.strip()), int(end.strip().replace("+", ""))
    else:
        week = int(timeline.strip().replace("+", ""))
        return week, week

change_phases = list(change_management.keys())
change_timelines = [extract_weeks(change_management[phase]['timeline']) for phase in change_phases]

# Process deployment strategy durations
deployment_phases = list(deployment_strategy.keys())
phase_durations = [int(deployment_strategy[phase]["duration"].split()[0]) for phase in deployment_phases]
cumulative_weeks = []
total_weeks = 0
for duration in phase_durations:
    cumulative_weeks.append((total_weeks, total_weeks + duration))
    total_weeks += duration

# Plot both
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10))

# Change management timeline
for i, (phase, (start, end)) in enumerate(zip(change_phases, change_timelines)):
    ax1.barh(i, end - start + 1, left=start - 1, height=0.6, alpha=0.7)
    ax1.text(start + (end - start) / 2 - 0.5, i, phase, ha='center', va='center', fontweight='bold', fontsize=9)
ax1.set_yticks(range(len(change_phases)))
ax1.set_yticklabels(change_phases)
ax1.set_xlabel('Project Week')
ax1.set_title('Change Management Timeline (ADKAR Model)', fontweight='bold')
ax1.grid(True, alpha=0.3)

# Deployment timeline
colors = ['lightblue', 'lightgreen', 'lightcoral', 'lightgray']
for i, (phase, (start, end)) in enumerate(zip(deployment_phases, cumulative_weeks)):
    ax2.barh(i, end - start, left=start, height=0.6, alpha=0.8, color=colors[i % len(colors)])
    ax2.text(start + (end - start) / 2, i, f'{phase}\n({end - start}w)', ha='center', va='center', fontweight='bold', fontsize=9)
ax2.set_yticks(range(len(deployment_phases)))
ax2.set_yticklabels([p.replace('Phase ', 'P') for p in deployment_phases])
ax2.set_xlabel('Deployment Week')
ax2.set_title('Phased Deployment Strategy', fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
